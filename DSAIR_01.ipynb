```python
class BankAccount():
    account_number = 101
    # initializer,
    # automatically executed when you create an object.
    def __init__(self, name, amount):
        self.name = name
        self.acc_no = BankAccount.account_number
        BankAccount.account_number += 1
        self.__amount = amount # private variable.
        print(name + "'s new account opened")

    def __repr__(self):
        return f"{self.name}, with account num: {self.acc_no} has Rs.{self.__amount}"

    def get_balance(self):
        return f"Your balace Rs.{self.__amount}"

    def withdrawl(self, amt):
        if amt > self.__amount:
            return "Insufficient Funds"

        self.__amount -= amt
        print(f"Withdrawing cash Rs.{amt} krrrrrrrrr...")
        print(f"Balance remaining Rs.{self.__amount}")

    def deposit(self, amt):
        if amt <=0:
            return "Put some money here"

        self.__amount += amt
        print(f"Deposited cash Rs.{amt}")
        print(f"Updated balance Rs.{self.__amount}")

    def __add__(self, other):
        new_name = f"{self.name}+{other.name}"
        new_balance = self.__amount + other.__amount
        return BankAccount(new_name, new_balance)

     def __lt__(self, other):
         return self.__amount < other.__amount

    def __eq__(self, other):
         return self.__amount == other.__amount
```

```python
class SavingsAccount(BankAccount):
    def __init__(self, name, amount):
        super().__init__(name, amount) # calling the parent __init__
        self.__amount = amount
        self.intreset_rate = 0.035
        self.max_transc_limit = 10
        self.current_trnsc = 0

    def calculate_intreset(self):
        intrest = self.__amount * self.intreset_rate
        return f"Your total interest Rs. {intrest:0.2f}"
```

```python
mohit_sv = SavingsAccount("Mohit", 10000)
```

```python
print(mohit_sv)
```

```python
mohit_sv.calculate_intreset()
```

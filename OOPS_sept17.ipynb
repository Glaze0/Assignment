```python
class Playlist():
    def __init__(self, name):
        self.name = name
        self.songs = []

    def add_song(self, song):
        self.songs.append(song)

    def show_playlist(self):
        print(f"🎵 {self.name}")
        for song in self.songs:
            print(f"  - {song}")

    def num_songs(self):
        return len(self.songs)

    def duration(self):
        total_duration = 0
        for song in self.songs:
            total_duration += song.duration


        return f"Total Duration {self.name} : {total_duration//60} mins {total_duration%60} seconds"
```

## Build a Swiggy Cart using :

- A MenuItem class with name, price, and a __repr__
- A Cart class that has a list of MenuItem objects, with add(item), remove(name), and total() methods
- A GoldCart(Cart) child class that overrides total() to apply a 10% discount
- Print a receipt for both a normal cart and a Gold cart holding the same items

```python
class MenuItem():
  def __init__(self,name,price):
    self.name = name
    self.price = price

  def __repr__(self):
    return f"{self.name}:{self.price}"
```

```python
class cart():
  def __init__(self) -> None:
    self.items = []


  def add(self,item):
    self.items.append(item)
  def remove(self,name):
    self.items.remove(name)
  def total(self):
    total = 0
    for item in self.items:
      total += item.price
    return total
  def show_cart(self):
    return f"Cart: {self.items}"
```

```python
class GoldCart(cart):
  def total(self):
    total = 0
    for item in self.items:
      total += item.price
    return total*0.9
```

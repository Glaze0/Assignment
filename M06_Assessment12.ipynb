```python
def build_augmented_prompt(chunks, question):
    instruction = (
        "You're a helpful assistant. Answer the question using the provided context. "
        "If the answer is not available, say I do not know."
    )

    content = ''
    for chunk  in chunks:
      content = content +'\n\n' + chunk

    prompt = (
        f'''{instruction}\n
Context: {content}\n
Question: \n{question}
        '''
    )

    return prompt


if __name__ == "__main__":
    sample_chunks = [
        "Refunds are processed within 5-7 business days after approval.",
        "Store credit is issued immediately upon return confirmation."
    ]
    sample_question = "How long does a refund take?"
    print(build_augmented_prompt(sample_chunks, sample_question))
```

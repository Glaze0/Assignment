#Scaling
Min-Max scaling for a large amount of data, value ranges from (0-1) used when calcualting mean and standard deviation is hard.

Formula: `x-min/max-min`.

and use z-score(standard scaling) when data is not that big.

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID":[101,102,103,104,105],
    "Gender":["Male","Female","Female","Male","Female"],
    "City":["Pune","Mumbai","Delhi","Pune","Mumbai"],
    "Age":[25,None,35,40,30],
    "Purchase_Amount":[1200,800,2500,1800,900],
    "MRP":[1500,1000,3000,2000,1000],
    "Joining_Date":[
        "2022-01-15",
        "2021-06-10",
        "2023-02-20",
        "2020-09-05",
        "2022-12-12"
    ],
    "Purchased":[1,0,1,1,0]
})

df
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["Gender"] = encoder.fit_transform(df["Gender"])

df
```

#Filling Missing data

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

#Scaling

```python
# 1. Min Max Scaling --> Normalization

# x --> (x-min)/(max - min)

from sklearn.preprocessing import MinMaxScaler

mn_mx_sc = MinMaxScaler()

df["Age_scaled"] = mn_mx_sc.fit_transform(df[["Age"]])
df
```

```python
#2. Standardization --> Standard Scaler --> mean is close to 0 and std is close to 1
```

```python
from sklearn.preprocessing import StandardScaler

std_sc = StandardScaler()

df["MRP_scaled"] = std_sc.fit_transform(df[["MRP"]])
df
```

#Feature Engineering

```python
#Brief cleaning for recommendation use-case
df['InvoiceData'] = pd.to_datetime(df['InvoiceDate'], format='%m/%d/%y %H:%M', errors='coerce')
df = df.dropna(subset=['CustomerID','Description','InvoiceDate'])

#Remove cancellations
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]

df['CustomerID'] = df['CustomerID'].astype(int)
df['Revenue'] = df['Quantity'] * df['UnitPrice']

print('Cleaned shape:', df.shape)
print(f'Customers:{df['CustomerID'].nunique()} | Products:{df['StockCode'].nunique()}')
```

#1) User Behaviour & Purchase Pattern Analysis
We start by answering:
* When do users buy?

```python
# Purchase pattern by month and hours
df['month'] = df['InvoiceData'].dt.to_period('M').astype(str)
df['hour'] = df['InvoiceData'].dt.hour

monthly_orders = df.groupby('month')['InvoiceNo'].nunique()
hourly_orders = df.groupby('hour')['InvoiceNo'].nunique()

top_products = (
    df.groupby('Description', as_index=False)
    .agg(total_qty=('Quantity', 'sum'), total_rev = ('Revenue','sum'))
    .sort_values('total_qty', ascending=False)
    .head(10)
)

fig, axes = plt.subplots(1,2, figsize=(15,5))
monthly_orders.plot(kind='line', marker='o',ax = axes[0], title='Monthly Orders')
hourly_orders.plot(kind ='bar', ax = axes[1], title='Hourly Orders')
axes[0].tick_params(axis = 'x', rotation = 45)
plt.tight_layout()
plt.show()

top_products
```

#2) Collaborative Filtering
Theory:
* **User-based CF:** Find users with similar purchase vectors, recommend what neighbours bought.
* **Iten-based CF:** Find items often co-purchased by similar users; usually more stable in production.

In many e-commerce, item-based CF is easier to scale and refresh.

```python
#Build user-item implicit matrix
user_items = (
    df.groupby(['CustomerID','StockCode'])['Quantity']
    .sum()
    .unstack(fill_value=0)
)

#Keep active users/item for faster workshop execution
active_users = user_items.astype(bool).sum(axis = 1) >= 5
active_items = user_items.astype(bool).sum(axis = 0) >= 5
user_items = user_items.loc[active_users, active_items]

print(f'Matrix shape: {user_items.shape}')

# Similarity Matrices
user_sim = pd.DataFrame(
    cosine_similarity(csr_matrix(user_items.values)),
    index = user_items.index,
    columns = user_items.index
)

item_sim = pd.DataFrame(
    cosine_similarity(csr_matrix(user_items.values.T)),
    index = user_items.columns,
    columns=user_items.columns
)
```

```python
def recommend_item_cf(customer_id, top_n=10):
  if customer_id not in user_items.index:
    return pd.DataFrame({'message':['Customer not in training matrix']})

  purchased = user_items.loc[customer_id]
  purchased_items = purchased[purchased > 0].index

  scores = pd.Series(0.0, index = user_items.columns)
  for item in purchased_items:
    scores = scores.add(item_sim[item] * purchased[item], fill_value = 0)

  scores = scores.drop(labels = purchased_items, errors = 'ignore').sort_values(ascending = False).head(top_n)

  name_map = df[['StockCode', 'Description']].drop_duplicates('StockCode').set_index('StockCode')['Description']
  return pd.DataFrame({
      'StockCode': scores.index,
      'item_cf_score': scores.values,
      'Description': name_map.reindex(scores.index).values
  })


sample_customer = int(user_items.index[0])
recommend_item_cf(sample_customer, top_n=8)
```

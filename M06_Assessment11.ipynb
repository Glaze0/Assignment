```python
SAMPLE_DOC = """RAG systems combine retrieval with generation. They fetch relevant chunks from a vector store before answering.

Chunking quality directly affects retrieval quality. A chunk that mixes two unrelated ideas is hard to retrieve correctly. Good chunks stay focused on one topic.

Overlap between chunks preserves context across boundaries."""

MAX_SIZE = 120

def split_sentences(paragraph):
    raw_sentences = [s.strip() for s in paragraph.split('.') if s.strip()]
    sentences = []
    for i, s in enumerate(raw_sentences):
      #Adding back the trailing '.'
      if i < len(raw_sentences) - 1 or paragraph.strip().endswith('.'):
        sentences.append(s + '.')
      else:
        sentences.append(s)

    return sentences

def chunk_document(text, max_size):
    paragraphs = [p.strip() for p in text.split("\n\n") if p.strip()]
    chunks = []
    for para in paragraphs:
        if len(para) <= max_size:
            chunks.append(para)
            continue
        # TODO: split `para` into sentences and greedily combine
        # consecutive sentences into chunks of at most `max_size`
        # characters, appending each finished chunk to `chunks`.
        sen = split_sentences(para)
        current_chunk = ''
        for s in sen:
          if current_chunk:
            current_chunk = s
          elif len(current_chunk) + len(s) <= max_size:
            current_chunk += s
          else:
            chunks.append(current_chunk)
            current_chunk = s
        if current_chunk:
          chunks.append(current_chunk)

    return chunks

if __name__ == "__main__":
    print(chunk_document(SAMPLE_DOC, MAX_SIZE))
```

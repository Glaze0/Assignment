```python
# ============================================================
# 1. INSTALL THE GEMINI PYTHON SDK
# ============================================================

# The Google GenAI SDK provides Python functionality
# for interacting with Gemini models.
#
# -q = quiet installation
# -U = upgrade to the latest version

!pip install -q -U google-genai


# ============================================================
# 2. IMPORT THE GEMINI LIBRARY
# ============================================================

from google import genai


# ============================================================
# 3. ADD YOUR API KEY
# ============================================================

# API keys authenticate our application when
# making requests to the Gemini API.
#
# For this classroom exercise, we are placing
# the API key directly in the code.
#
# IMPORTANT:
# Do not share your API key publicly.
# In production, use environment variables,
# Colab Secrets, or a secret manager.

API_KEY = "__"
```

```python
# ============================================================
# 4. CREATE THE GEMINI CLIENT
# ============================================================

# The Gemini client manages communication between
# our Python application and the Gemini API.

client = genai.Client(api_key=API_KEY)

print("Gemini client created successfully!")


# ============================================================
# 5. MAKE YOUR FIRST GEMINI API CALL
# ============================================================

# generate_content() sends a request to the
# specified Gemini model.
#
# model:
#     Specifies which Gemini model to use.
#
# contents:
#     Contains the prompt sent to Gemini.
#
# response:
#     Stores the response returned by Gemini.
#
# response.text:
#     Extracts the generated text.

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="Explain Generative AI in simple terms."
)

print(response.text)

#+++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++
# You can also use something like 'client.chats.create(model='X')' to make an
# object and then use 'object.send_message('prompt')' to communicate with Gemini


from google import genai

API_KEY = "XX"

client = genai.Client(api_key=API_KEY)

chat = client.chats.create(
    model="gemini-3.5-flash"
)

# Now you can keep sending messages without repeating model/AFC
resp = chat.send_message(
            "Explain AI in 3 lines?"
)

resp = chat.send_message(message='can you explain me about motors in 3 lines',)

print(resp.text)




# ============================================================
# 6. CHANGE THE PROMPT
# ============================================================

# The API code remains the same.
# Only the prompt changes.

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="Explain RAG to a beginner using a restaurant analogy."
)

print(response.text)


# ============================================================
# KEY CONCEPT
# ============================================================
#
# Same API
#     +
# Different Prompt
#     =
# Different Task
#
# The API integration remains the same while
# the prompt determines what Gemini needs to do.


# ============================================================
# 7. CREATE A REUSABLE PYTHON FUNCTION
# ============================================================

# Instead of writing the complete API call every time,
# we can place it inside a reusable function.

def ask_gemini(prompt):

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )

    return response.text


# ============================================================
# 8. USE THE FUNCTION
# ============================================================

# We can now simply pass a prompt to our function.

print(ask_gemini("What is an LLM?"))


# ============================================================
# WHY USE A FUNCTION?
# ============================================================
#
# Without a function:
#
# client.models.generate_content(...)
# client.models.generate_content(...)
# client.models.generate_content(...)
#
#
# With a function:
#
# ask_gemini("What is an LLM?")
# ask_gemini("What is RAG?")
# ask_gemini("What are embeddings?")
#
#
# Benefits:
# - Reusable
# - Cleaner code
# - Easier to maintain
# - Easier to integrate into applications


# ============================================================
# 9. TEST MULTIPLE PROMPTS
# ============================================================

print("\n--- QUESTION 1 ---")
print(ask_gemini("What is an embedding?"))

print("\n--- QUESTION 2 ---")
print(ask_gemini("What is Retrieval Augmented Generation?"))

print("\n--- QUESTION 3 ---")
print(ask_gemini("What is an AI agent?"))


# ============================================================
# 10. DYNAMIC PROMPTS USING PYTHON
# ============================================================

# Real-world AI applications usually construct
# prompts dynamically.
#
# Python variables can be inserted into prompts
# using formatted-strings.

topic = "Generative AI"

prompt = f"""
Explain {topic} to a beginner.

Give me:
1. A simple definition
2. One real-world example
3. One limitation
"""

print(ask_gemini(prompt))


# ============================================================
# 11. CHANGE THE PYTHON VARIABLE
# ============================================================

# We can change the topic without changing
# the overall prompt structure.

topic = "Retrieval Augmented Generation"

prompt = f"""
Explain {topic} to a beginner.

Give me:
1. A simple definition
2. One real-world example
3. One limitation
"""

print(ask_gemini(prompt))


# ============================================================
# DYNAMIC PROMPTING
# ============================================================
#
# Python Variable
#       ↓
# Prompt Construction
#       ↓
# Gemini API
#       ↓
# Gemini Model
#       ↓
# Response
#
# This is an important pattern for building
# real-world LLM applications.


# ============================================================
# 12. ANOTHER DYNAMIC PROMPT EXAMPLE
# ============================================================

topic = "Machine Learning"
audience = "college students"

prompt = f"""
Explain {topic} to {audience}.

Requirements:
- Use simple language
- Give one real-world example
- Keep the answer below 150 words
"""

print(ask_gemini(prompt))


# ============================================================
# FINAL RECAP
# ============================================================
#
# We learned how to:
#
# 1. Install the Gemini Python SDK
# 2. Import the SDK
# 3. Provide an API key
# 4. Create a Gemini client
# 5. Call generate_content()
# 6. Specify a Gemini model
# 7. Send a prompt
# 8. Read response.text
# 9. Create a reusable Python function
# 10. Build dynamic prompts using Python
#
#
# CORE PATTERN:
#
# Python
#    ↓
# Prompt
#    ↓
# Gemini API
#    ↓
# Gemini Model
#    ↓
# Response
#    ↓
# Python Application
#
#
# KEY TAKEAWAY:
#
# Gemini API + Python allows us to integrate
# generative AI capabilities into our own
# Python applications.
```

```python
# ============================================================
# BLOCK 2 — GEMINI 3.5 CONFIGURATION
# ============================================================
#
# Topics:
# 1. Temperature
# 2. Top-P
# 3. Top-K
# 4. Max Output Tokens
# 5. Response MIME Type
# 6. Response Schema
# 7. Thinking Level
# 8. System Instructions
#
# Model:
# gemini-3.5-flash
#
# ============================================================


from google import genai
from google.genai import types


# ============================================================
# 1. TEMPERATURE
# ============================================================
#
# Temperature controls the randomness / variability of token selection.
#
# Lower temperature:
# → More predictable
# → More focused
#
# Higher temperature:
# → More varied
# → More creative
#
# Typical examples:
#
# 0.2 → more focused
# 0.7 → balanced
# 1.0 → more varied
#
#
# ============================================================
# TEMPERATURE EXAMPLE
# ============================================================

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="""
    Give me 5 creative names for an AI startup.
    """,
    config=types.GenerateContentConfig(
        temperature=0.7
    )
)

print(response.text)


# Try changing:
#
# temperature=0.1
#
# and then:
#
# temperature=1.0
#
# Compare the outputs.
```

```python
# ============================================================
# 2. TOP-K
# ============================================================
#
# Top-K controls how many of the most probable
# candidate tokens are considered.
#
# Example:
#
# top_k=10
#
# → Consider the top 10 candidate tokens.
#
# Smaller top_k:
# → Smaller candidate pool
#
# Larger top_k:
# → Larger candidate pool
#
#
# ============================================================
# TOP-K EXAMPLE
# ============================================================

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="""
    Give me 5 creative names for an AI startup.
    """,
    config=types.GenerateContentConfig(
        top_k=10
    )
)

print(response.text)


# Try:
#
# top_k=1
# top_k=10
# top_k=40
#
# and compare the outputs.


# ============================================================
# 3. TOP-P
# ============================================================
#
# Top-P is also called nucleus sampling.
#
# Instead of selecting a fixed number of tokens,
# top-p selects tokens based on cumulative probability.
#
# Example:
#
# top_p=0.9
#
# → Consider the most probable tokens until their
#   cumulative probability reaches approximately 0.9.
#
# Lower top_p:
# → More restricted candidate pool
#
# Higher top_p:
# → Broader candidate pool
#
#
# ============================================================
# TOP-P EXAMPLE
# ============================================================

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="""
    Give me 5 creative names for an AI startup.
    """,
    config=types.GenerateContentConfig(
        top_p=0.9
    )
)

print(response.text)


# Try:
#
# top_p=0.5
# top_p=0.9
# top_p=0.95
#
# and compare the outputs.


# ============================================================
# 4. USING TEMPERATURE + TOP-P + TOP-K
# ============================================================
#
# These parameters can be specified together.
#
# Temperature
# → Controls randomness
#
# Top-K
# → Limits the candidate pool by number
#
# Top-P
# → Limits the candidate pool by probability
#
#
# ============================================================

response = client.models.generate_content(
    model="gemini-3.5-flash",

    contents="""
    Create a creative advertisement for an
    AI-powered education platform.
    """,

    config=types.GenerateContentConfig(
        temperature=0.8,
        top_p=0.9,
        top_k=40
    )
)

print(response.text)


# ============================================================
# IMPORTANT GEMINI 3.5 NOTE
# ============================================================
#
# Although these parameters can be demonstrated,
# Google currently recommends NOT changing them for
# Gemini 3.x models.
#
# Gemini 3.x reasoning is optimized around the
# default generation settings.
#
# Therefore:
#
# FOR LEARNING:
# → Experiment with them to understand the concepts.
#
# FOR PRODUCTION GEMINI 3.5:
# → Prefer the model defaults.
#
# Google specifically recommends removing these
# parameters from Gemini 3.x requests.
#
#
# ============================================================
# 5. MAX OUTPUT TOKENS
# ============================================================
#
# max_output_tokens specifies the maximum number
# of tokens Gemini can generate.
#
# Useful for:
# - Short answers
# - Summaries
# - Controlling response size
# - Application constraints
#
#
# ============================================================

response = client.models.generate_content(
    model="gemini-3.5-flash",

    contents="""
    Explain Generative AI to a beginner.
    """,

    config=types.GenerateContentConfig(
        max_output_tokens=100
    )
)

print(response.text)


# ============================================================
# 6. RESPONSE MIME TYPE
# ============================================================
#
# response_mime_type specifies the format of the generated response.
#
# text/plain
# → Normal text
#
# application/json
# → JSON
#
#
# ============================================================
# NORMAL TEXT
# ============================================================

response = client.models.generate_content(
    model="gemini-3.5-flash",

    contents="What is an API?",

    config=types.GenerateContentConfig(
        response_mime_type="text/plain"
    )
)

print(response.text)


# ============================================================
# JSON RESPONSE
# ============================================================

response = client.models.generate_content(
    model="gemini-3.5-flash",

    contents="""
    Give me a fictional person's name and age.
    Return the result as JSON.
    """,

    config=types.GenerateContentConfig(
        response_mime_type="application/json"
    )
)

print(response.text)


# Example:
#
# {
#   "name": "Rahul Sharma",
#   "age": 28
# }


# ============================================================
# 7. RESPONSE SCHEMA
# ============================================================
#
# response_mime_type tells Gemini:
#
# "Return JSON."
#
# response_schema tells Gemini:
#
# "Return JSON with THIS structure."
#
#
# ============================================================

response = client.models.generate_content(
    model="gemini-3.5-flash",

    contents="""
    Give me a fictional person's name and age.
    """,

    config=types.GenerateContentConfig(

        response_mime_type="application/json",

        response_schema={
            "type": "OBJECT",

            "properties": {

                "name": {
                    "type": "STRING"
                },

                "age": {
                    "type": "INTEGER"
                }
            },

            "required": [
                "name",
                "age"
            ]
        }
    )
)

print(response.text)


# Expected structure:
#
# {
#   "name": "Rahul Sharma",
#   "age": 28
# }


# ============================================================
# 8. THINKING LEVEL
# ============================================================
#
# Gemini 3.5 provides thinking_level to control reasoning effort.
#
# Available values:
#
# minimal
# low
# medium
# high
#
#
# Simple task:
# → minimal / low
#
# General task:
# → medium
#
# Complex reasoning:
# → high
#
#
# ============================================================
# SIMPLE TASK
# ============================================================

response = client.models.generate_content(
    model="gemini-3.5-flash",

    contents="Define machine learning in one sentence.",

    config=types.GenerateContentConfig(
        thinking_config=types.ThinkingConfig(
            thinking_level="minimal"
        )
    )
)

print(response.text)


# ============================================================
# COMPLEX TASK
# ============================================================

response = client.models.generate_content(
    model="gemini-3.5-flash",

    contents="""
    A company receives 10,000 customer support tickets.

    40% are billing issues.
    30% are technical issues.
    20% are account issues.
    The remaining tickets are general queries.

    Calculate the number of tickets in each category.
    Explain your reasoning.
    """,

    config=types.GenerateContentConfig(
        thinking_config=types.ThinkingConfig(
            thinking_level="high"
        )
    )
)

print(response.text)


# ============================================================
# 9. SYSTEM INSTRUCTION
# ============================================================
#
# System instruction tells the model HOW to behave.
#
# User prompt:
# → WHAT should the model do?
#
# System instruction:
# → HOW should the model behave?
#
#
# ============================================================

response = client.models.generate_content(
    model="gemini-3.5-flash",

    contents="Explain Python decorators.",

    config=types.GenerateContentConfig(

        system_instruction="""
        You are an expert Python instructor.

        Teach beginners using simple language.

        Always:
        - Give a simple definition
        - Give one practical example
        - Explain why it is useful
        - Avoid unnecessary jargon
        """
    )
)

print(response.text)


# ============================================================
# 10. COMBINING CONFIGURATION OPTIONS
# ============================================================
#
# We can combine multiple configuration settings
# in one request.
#
# Example:
#
# System instruction
# +
# Thinking level
# +
# Output limit
# +
# JSON response
# +
# Response schema
#
#
# ============================================================

customer_message = """
I have been trying to reset my password for two hours.
The reset email never arrives and I urgently need access
to my account.
"""


response = client.models.generate_content(

    model="gemini-3.5-flash",

    contents=f"""
    Analyze this customer support message:

    {customer_message}

    Identify:
    - category
    - sentiment
    - priority
    """,

    config=types.GenerateContentConfig(

        system_instruction="""
        You are a customer support triage assistant.

        Analyze the customer's message carefully.
        Do not invent information.
        Return only the requested information.
        """,

        thinking_config=types.ThinkingConfig(
            thinking_level="medium"
        ),

        max_output_tokens=200,

        response_mime_type="application/json",

        response_schema={
            "type": "OBJECT",

            "properties": {

                "category": {
                    "type": "STRING"
                },

                "sentiment": {
                    "type": "STRING"
                },

                "priority": {
                    "type": "STRING"
                }
            },

            "required": [
                "category",
                "sentiment",
                "priority"
            ]
        }
    )
)

print(response.text)


# ============================================================
# FINAL QUICK REFERENCE
# ============================================================
#
# TEMPERATURE
# → Controls randomness / variability
#
# TOP-K
# → Controls number of candidate tokens
#
# TOP-P
# → Controls cumulative probability of candidates
#
# MAX_OUTPUT_TOKENS
# → Limits generated output
#
# RESPONSE_MIME_TYPE
# → Controls output format
#
# RESPONSE_SCHEMA
# → Defines the structure of the output
#
# THINKING_LEVEL
# → Controls reasoning effort
#
# SYSTEM_INSTRUCTION
# → Controls model behavior
#
#
# ============================================================
# IMPORTANT GEMINI 3.5 TAKEAWAY
# ============================================================
#
# Temperature / Top-P / Top-K:
# → Learn the concepts
# → They can be supplied in the API
# → But Google recommends keeping their defaults
#   for Gemini 3.x
#
# Thinking Level:
# → Current Gemini 3.x control for reasoning effort
#
# Structured Output:
# → response_mime_type + response_schema
#
# System Instruction:
# → Defines how the model should behave
#
# Max Output Tokens:
# → Controls maximum generated output
```

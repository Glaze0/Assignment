```python
import math

CHUNK_VECTORS = {
    0: [1, 8, 1, 0],
    1: [8, 1, 1, 2],
    2: [0, 8, 2, 0],
    3: [7, 2, 1, 3],
}
CHUNK_TEXTS = {
    0: "Standard shipping takes 5 to 7 business days within the country.",
    1: "Refunds are issued to the original payment method within 10 business days.",
    2: "Express shipping takes 1 to 2 business days and costs an extra fee.",
    3: "Refund requests must be submitted within 30 days of the purchase date.",
}
QUERY_TEXT = "How long does it take to get a refund back?"
QUERY_VECTOR = [9, 1, 0, 2]


def cosine_similarity(vec1, vec2):
    dot = sum(a*b for a, b in zip(vec1, vec2))
    mag1 = math.sqrt(sum(a*a for a in vec1))
    mag2 = math.sqrt(sum(b*b for b in vec2))
    return dot / (mag1 * mag2)


def retrieve(query_vector, chunk_vectors, chunk_texts, k=2):
    scored = []
    for idx, vec in chunk_vectors.items():
        score = cosine_similarity(query_vector, vec)
        scored.append((score, chunk_texts[idx]))
    scored.sort(key=lambda x: x[0], reverse=True)
    return scored[:k]


def build_prompt(query_text, retrieved_chunks):
    lines = []
    lines.append("You are a helpful assistant. Answer using only the context below.")
    lines.append("If the answer isn't there, say you don't know.\n")
    lines.append("Context:")
    for i, (score, text) in enumerate(retrieved_chunks, start=1):
        lines.append(f"[{i}] {text}")
    lines.append(f"\nQuestion: {query_text}")
    return "\n".join(lines)


if __name__ == "__main__":
    top_chunks = retrieve(QUERY_VECTOR, CHUNK_VECTORS, CHUNK_TEXTS, k=2)
    for score, text in top_chunks:
        print(f"{score:.3f} {text}")
    print()
    print(build_prompt(QUERY_TEXT, top_chunks))
```

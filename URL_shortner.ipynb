```python
from os import error
import pyshorteners
long_url = input("Enter the URL to shorten: ")



try:
  tiny_url = pyshorteners.Shortener().tinyurl.short(long_url)

  print("The Shortened URL is: " + tiny_url)

except(error):
  print('Operation Failed. ')
```

#Bitly


```python
import pyshorteners
long_url = input("Enter the URL to shorten: ")


apikey = input("Enter your API key: ")

#Bitly shortener service
type_bitly = pyshorteners.Shortener(api_key=apikey)
short_url = type_bitly.bitly.short(long_url)

print("The Shortened URL is: " + short_url)
```
